```python
import requests
from bs4 import BeautifulSoup
from fake_useragent import UserAgent
import csv
print('Your enivornment is ready to go')
```

```python
ua = UserAgent()
target_url = "https://www.amazon.com/s?k=good+quality+bluetooth+headphones"

entry_headers = {
    "User-Agent":ua.random,
    "Accept-Language":"en-UAE,en-US;q=0.8,en;q=0.8"
}
responce = requests.get(target_url,headers=entry_headers)
print(responce.status_code)
```

```python
raw_html = responce.text
```

```python
soup = BeautifulSoup(raw_html,"html.parser")
headphones = soup.find_all("div",class_="puis-card-container")
```

```python
for index,headphone in enumerate(headphones,1):
    title = headphone.find("h2")
    price = headphone.find("span",class_="a-price-whole")
    rating = headphone.find("i",class_="a-icon")
    if title and price and rating:
        if rating == 3:
            print('Has low rating')
        else:
            print(f"{index}.Title: {title.text} |\nPrice: {price.text} |\nRating: {rating.text}")
    else:
        continue
```

```python
with open("headphones.csv","w",encoding="utf-8") as my_file:
    writer = csv.writer(my_file)
    writer.writerow(['Index','Title','Price','Rating'])
    for index,headphone in enumerate(headphones,1):
        title = headphone.find("h2")
        price = headphone.find("span",class_="a-price-whole")
        rating = headphone.find("i",class_="a-icon")
        if title and price and rating:
            if rating == 3:
                print('Has low rating')
            else:
                writer.writerow([index,
                                 title.text.strip(),
                                 price.text,
                                 rating.text])
        else:
            continue
print("Your data successfully loaded 🫡")
    
    
```
